```python
df = df_sales.merge(
    df_customers,
    how='left',
    left_on='Customer Name Index',
    right_on='Customer Index'
)
```

```python
df.head(5)
```

```python
 #merge with products
 df = df.merge(
     df_products,
     how='left',
     left_on='Product Description Index',
     right_on='Index'
 )
```

```python
df.head(5)
```

```python
  df = df.merge(
    df_regions,
    how='left',
    left_on='Delivery Region Index',
    right_on='id'
)
```

```python
df.head(5)
```

```python
import re
df.columns = [re.sub(r'[^a-zA-Z0-9_]', '', col).lower() for col in df.columns]
# print(df.columns.tolist()) # Debugging line - removed after identifying the issue
df = df.merge(
        df_state_reg[['State Code', 'Region']],
        how='left',
        left_on='state_code',
        right_on='State Code'
    )
```

```python
df = df.merge(
    df_budgets,
    how='left',
    left_on='productname',
    right_on='Product Name'
)
```

```python
df.to_csv('file.csv')
```

```python
cols_to_drop=['Customer Index','Index','id','State Code']
df=df.drop(columns=cols_to_drop,errors='ignore')
df.head(5)
```

```python
#convert all columns to lower case
df.columns = df.columns.str.lower()
df.columns.values
```

```python
cols_to_keep = [
    'ordernumber',
    'orderdate',
    'customernames',
    'channel',
    'product name',
    'orderquantity',
    'unitprice',
    'linetotal',
    'totalunitcost',
    'state_code',
    'county',
    'state',
    'region',
    'latitude',
    'longitude',
    '2017 budgets' # Added to keep the merged budget column
]
```

```python
df = df[cols_to_keep]
```

```python
df.head(5)
```

```python
df = df.rename(columns={
   'ordernumber': 'order_number',
   'orderdate': 'order_date',
    'customernames': 'customer_name',
    'productname_x': 'product_name',
    'orderquantity': 'order_quantity',
    'unitprice': 'unit_price',
    'linetotal': 'revenue',
    'totalunitcost': 'cost',
    'state_code_y': 'state',
    'latitude_y': 'lat',
    'longitude_y': 'lon',
    '2017 budgets': 'budget' # Renamed after being correctly merged and lowercased
})
df.head(1)
```

```python
#blank out budgets for non-2017 orders
print('Columns before operation:', df.columns.tolist())
print('DataFrame head before operation:')
print(df.head())
df.loc[df['order_date'].dt.year != 2017, 'budget'] = pd.NA
df[['order_date','product name','revenue','budget']].head(5)
```
